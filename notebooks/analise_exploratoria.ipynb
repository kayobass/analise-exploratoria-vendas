```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('../data/vendas.csv')
```

```python
df.shape
```

```python
df.describe()
```

```python
receita_produto = df.groupby('produto')['receita'].sum().sort_values(ascending=False)
receita_produto
```

```python
plt.bar(receita_produto.index, receita_produto.values, color='purple')

plt.title('Receita total por Produto')
plt.xlabel('Produto')
plt.ylabel('Receita Total (R$)')
plt.grid(axis='y')

plt.show()
```

```python
qtd_produto = df.groupby('produto')['quantidade'].sum().sort_values(ascending=False)

display(receita_produto)
qtd_produto
```

```python
receita_categoria = df.groupby('categoria')['receita'].sum().sort_values(ascending=False)
receita_categoria
```

```python
plt.bar(receita_categoria.index, receita_categoria.values)

plt.title("Receita Total por Categoria")
plt.xlabel("Categorias")
plt.ylabel("Receita Total (R$)")
plt.grid(axis='y')

plt.show()
```

```python
df['data'] = pd.to_datetime(df['data'], format='%Y-%m-%d')
df.head()
```

```python
df['mes_dia'] = df['data'].dt.strftime('%m-%d')
receita_diaria = df.groupby('mes_dia')['receita'].sum()
receita_diaria
```

```python
plt.plot(receita_diaria.index, receita_diaria.values, marker='s', color='green')

plt.title("Evolução da Receita Diária")
plt.xlabel("Data")
plt.ylabel("Receita Diária (R$)")
plt.grid()

plt.show()
```
